```python
import hashlib
import time
from datetime import datetime, timezone
from email.utils import parsedate_to_datetime
from http.client import IncompleteRead
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.request import urlopen

def download_checked(url, path, expected_hash, algorithm='sha256', expected_bytes=None, max_attempts=5):
    path = Path(path)
    if max_attempts < 1:
        raise ValueError('max_attempts must be positive')
    if path.exists():
        content = path.read_bytes()
    else:
        for attempt in range(1, max_attempts + 1):
            retry_after = None
            try:
                with urlopen(url, timeout=90) as response:
                    content = response.read()
                break
            except HTTPError as error:
                status = error.code
                retry_after = error.headers.get('Retry-After') if error.headers else None
                error.close()
                if status not in {408, 429, 500, 502, 503, 504}:
                    raise
                failure = f'HTTP {status}'
            except (URLError, TimeoutError, ConnectionError, IncompleteRead) as error:
                failure = type(error).__name__
            if attempt == max_attempts:
                raise RuntimeError(
                    f'Download failed for {path.name}: {failure} after {max_attempts} attempts. '
                    'Rerun the acquisition cell in this runtime to reuse verified cached clips.'
                ) from None
            delay = min(60.0, float(2 ** attempt))
            if retry_after:
                try:
                    server_delay = float(retry_after)
                except ValueError:
                    try:
                        server_delay = (parsedate_to_datetime(retry_after) - datetime.now(timezone.utc)).total_seconds()
                    except (TypeError, ValueError, OverflowError):
                        server_delay = 0.0
                if server_delay > 60:
                    raise RuntimeError(
                        f'{path.name}: {failure}; server requests a wait of {server_delay:.0f}s. '
                        'Wait that long, then rerun the acquisition cell; cached clips are preserved.'
                    ) from None
                delay = max(delay, server_delay)
            print(f'{path.name}: {failure}; retry {attempt + 1}/{max_attempts} in {delay:g}s', flush=True)
            time.sleep(delay)
    if hashlib.new(algorithm, content).hexdigest() != expected_hash:
        raise ValueError(f'Checksum mismatch: {path.name}; cached file was not overwritten')
    if expected_bytes is not None and len(content) != expected_bytes:
        raise ValueError(f'Unexpected size: {path.name}')
    if not path.exists():
        path.parent.mkdir(parents=True, exist_ok=True)
        temporary = path.with_suffix(path.suffix + '.part')
        temporary.write_bytes(content)
        temporary.replace(path)
    return content

print('Download helper ready: retry temporary HTTP failures; reuse checksum-verified cached clips.')
```

## 5. Acquire and prepare all training/validation inputs

This is the long one-time download/decode phase, before model training. It has
**4,304 clips**, with no 100-request or 256-clip training cap. Progress is printed

checks bind the source, preprocessing code, dimensions and decoder versions.

If HTTP acquisition fails, keep the runtime and rerun **this cell**: completed
clips are verified and reused. If preparation is interrupted, only an unfinished
clip is rebuilt. Existing corrupt committed files cause an explicit error. A new
Colab runtime needs the inputs downloaded/prepared again, but resumes saved model,
optimizer, scheduler, scaler, epoch cursor and RNG state from Drive afterward.


```python
if not unetpp_precision_passed:
    raise RuntimeError('Pass the real-input precision diagnostic first.')
unetpp_cache_identity = {'preprocessing': UNETPP_CONFIG['preprocessing'],
    'implementation': UNETPP_CONFIG['implementation_sha256']['unetpp_baseline'],
    'torch': str(torch.__version__), 'opencv': cv2.__version__}
unetpp_cache_signature = json_digest(unetpp_cache_identity)
unetpp_cache_dir = unetpp_data_dir/'prepared_unetpp'/unetpp_cache_signature
unetpp_cache_dir.mkdir(parents=True, exist_ok=True)
unetpp_sources = record_video_sources(unetpp_all, UNETPP_CONFIG['dataset_revision'],
    UNETPP_OUTPUT_DIR/'video_sources.json', HfApi(token=False))
unetpp_cache_required = len(unetpp_all)*96*128*128*4
unetpp_cache_existing = sum(path.stat().st_size for path in unetpp_cache_dir.glob('*.npy'))
unetpp_raw_missing = sum(unetpp_sources[row.file_path]['bytes'] for row in unetpp_all.itertuples(index=False)
    if not (unetpp_data_dir/'clips'/row.file_path).is_file())
unetpp_needed = max(0, unetpp_cache_required-unetpp_cache_existing)+unetpp_raw_missing+2_000_000_000
if shutil.disk_usage(unetpp_data_dir).free < unetpp_needed:
    raise RuntimeError(f'Need approximately {unetpp_needed/1e9:.1f} GB more runtime disk including safety margin; do not put the full cache on Drive.')
print('Full prepared cache:', round(unetpp_cache_required/1e9, 2), 'GB; source video bytes:',
    sum(unetpp_sources[value]['bytes'] for value in unetpp_all['file_path']), flush=True)
unetpp_paths, unetpp_checks = [], []
unetpp_prepare_start = time.perf_counter()
for index, row in enumerate(unetpp_all.itertuples(index=False), 1):
    path, check = prepare_cached_clip(row, unetpp_sources[row.file_path], unetpp_data_dir,
        unetpp_cache_dir, unetpp_cache_signature, download_checked)
    unetpp_paths.append(path)
    unetpp_checks.append(check)
    if index == 1 or index % 10 == 0 or index == len(unetpp_all):
        print(f'Hash-verified and prepared {index}/{len(unetpp_all)} clips', flush=True)
unetpp_cache_report = {'cache_identity': unetpp_cache_identity, 'clips': unetpp_checks,
    'manifest_hashes': unetpp_manifest_hashes, 'preparation_seconds_this_call': time.perf_counter()-unetpp_prepare_start,
    'test_used': False}
unetpp_data_fingerprint = json_digest({'cache_identity': unetpp_cache_identity,
    'clips': unetpp_checks, 'manifest_hashes': unetpp_manifest_hashes})
write_json_atomic(UNETPP_OUTPUT_DIR/'preprocessing.json', unetpp_cache_report)
unetpp_train_dataset = CachedClips(unetpp_paths[:len(unetpp_train)], unetpp_train['label'])
unetpp_validation_dataset = CachedClips(unetpp_paths[len(unetpp_train):], unetpp_validation['label'])
print('READY:', len(unetpp_train_dataset), 'train,', len(unetpp_validation_dataset), 'validation.')
print('Constant non-RD validation reference:', binary_metrics(unetpp_validation['label'].to_numpy(), np.zeros(len(unetpp_validation))))
```
